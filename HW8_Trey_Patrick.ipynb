```python
import requests

API_URL = "https://router.huggingface.co/hf-inference/models/sentence-transformers/all-MiniLM-L6-v2/pipeline/sentence-similarity"
headers = {"Authorization": "Bearer YOUR_HF_TOKEN"}

def query(payload):
    response = requests.post(API_URL, headers=headers, json=payload)
    return response.json()

files = ["essay1.txt", "essay2.txt", "essay3.txt"]
essays = [open(f, encoding="utf-8").read() for f in files]

for i in range(3):
    scores = query({"inputs": {"source_sentence": essays[i], "sentences": essays[i + 1:]}}) if i < 2 else []
    for j, score in zip(range(i + 1, 3), scores):
        flag = "RED FLAG" if score > 0.9 else "OK"
        print(f"{files[i]} vs {files[j]}: {score:.4f} {flag}")
```
