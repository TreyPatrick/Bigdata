## Task 1


```python
import requests

descriptions = []

for page in range(1, 4):
    url = "https://newsapi.org/v2/everything?q=OpenAI&sortBy=publishedAt&page=" + str(page) + "&apiKey=" + NEWS_API_KEY
    response = requests.get(url)
    json_data = response.json()

    articles = json_data['articles']
    for article in articles:
        descriptions.append(article['description'])

print(len(descriptions))
```

## Task 2


```python
headlines = []

for page in range(0, 5):
    url = "https://api.nytimes.com/svc/search/v2/articlesearch.json?api-key=" + NYT_API_KEY + "&q=OpenAI&sort=newest&page=" + str(page)
    response = requests.get(url)
    json_data = response.json()

    docs = json_data['response']['docs']
    for doc in docs:
        headlines.append(doc['headline']['main'])

print(len(headlines))
```
